# Creating Tensors  

```python
import torch
a = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype = torch.int32)
print(a)
```

```python
import torch
a = torch.linspace(-1, 1, 10)
print(a)
```

```python
import torch
a = torch.randint(3, 10, size=(3,4))
print(a)
```

```python
import torch
a = torch.zeros(2, 3, 4)
print(a)
```

```python
import torch
a = torch.full((2,3,4), 5)
print(a)
```

```python
import torch
a = torch.eye(4)
print(a)
```
