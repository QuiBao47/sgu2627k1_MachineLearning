### Bài 8. Autograd cơ bản
### Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.
### Yêu cầu:
### • Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.
### • Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

```python
x = torch.tensor(
3.6,
requires_grad=True      # buộc PyTorch phải ghi nhớ mọi phép toán liên quan đến biến này
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

```python
# 1. Tính thủ công bằng công thức toán học (y' = 2x)
x_val = 3.6
manual_grad = 2 * x_val

# 2. In kết quả để so sánh
print(f"Đạo hàm tính bằng PyTorch (x.grad): {x.grad.item()}")
print(f"Đạo hàm tính thủ công bằng code : {manual_grad}")
```

```python
# Khởi tạo tensor x
x = torch.tensor(3.6, requires_grad=True)

# Khai báo hàm số mới
y = 3 * x**2 + 2 * x + 1

# Tính đạo hàm tự động
y.backward()

print("x =", x.item())
print("y =", y.item())
print("x.grad (PyTorch tính) =", x.grad.item())
```

### Bài 9. Gradient Descent tìm hệ số đa thức
### Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.
### Yêu cầu:
### • So sánh ba hệ số học được với [1, 2, 3].
### • Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step(). (Làm vào word)

```python
import numpy as np
polynomial = np.poly1d([1, 2, 3])       # chức năng của thư viện numpy, np.poly1d([1, 2, 3]) tạo ra một hàm đa thức bậc 2 với hệ số [1, 2, 3], tương đương với f(x) = 1*x^2 + 2*x + 3
N = 20
X = np.random.randn(N, 1) * 5       # 1 là số lượng cột, tạo ra 20 số ngẫu nhiên được xếp thành 1 cột và được nhân với 5 để mở rộng phạm vi giá trị của các số ngẫu nhiên
Y = polynomial(X)       # gán 20 giá trị X vào hàm đa thức polynomial để tính ra 20 giá trị Y tương ứng
XX = np.hstack([        # tạo ra và ghép 3 cột lại tạo thành một ma trận có kích thước 20x3
    X * X,      # cột đầu tiên X được bình phương 
    X,      # cột thứ hai X giữ nguyên
    np.ones_like(X)     # cột thứ ba là một cột toàn số 1, có cùng kích thước với X
])
w = torch.randn(3, 1, requires_grad=True)       # Tạo ngẫu nhiên các giá trị, tạo tensor thành ma trận 3x1, buộc PyTorch phải ghi lại lịch sử mọi phép toán liên quan đến w để sau này có thể tự động tính đạo hàm (gradient) thông qua hàm .backward()
x = torch.tensor(XX, dtype=torch.float32)       # chuyển đổi ma trận NumPy này sang định dạng Tensor của PyTorch và ép kiểu số thực 32-bit
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)     # sử dụng thuật toán NAdam từ module optim của thư viện PyTorch.
print("Initial coefficients:")
print(w)        # kích thước ma trận 3 x 1
for epoch in range(1000):
    optimizer.zero_grad()       # Xóa sạch đạo hàm của vòng lặp cũ, tránh bị cộng dồn sai lệch
    y_pred = x @ w      # kích thước ma trận x: N x 3  
    mse = torch.mean(torch.square(y - y_pred))      # tính sai số bình phương trung bình lấy y trừ y_pred sau đó bình phương sai số và tính trung bình
    mse.backward()      # tính đạo hàm của sai số theo w
    optimizer.step()        # cập nhật w
print("Learned coefficients:")
print(w)
print("\n--- So sánh hệ số ---")
print("Hệ số gốc (đáp án)    : [1.0, 2.0, 3.0]")
print(f"Hệ số mô hình học được: [{w[0].item():.4f}, {w[1].item():.4f}, {w[2].item():.4f}]")

# Tính sai số (tùy chọn)
sai_so = torch.abs(w.squeeze() - torch.tensor([1.0, 2.0, 3.0]))     # .squeeze() dùng để loại bỏ các chiều có kích thước bằng 1.
print(f"Sai số tương đối      : {sai_so.detach().numpy()}")     # sai_so.detach() Tạo một Tensor mới tách khỏi đồ thị tính đạo hàm của PyTorch.
                                # Chuyển Tensor PyTorch thành mảng NumPy
```

### Bài 10. Giải hệ phương trình bằng Autograd
### Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.
### Yêu cầu:
### • Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
### • Nhận xét vì sao bài toán có thể có nhiều nghiệm. (Làm vào word)

```python
import random
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
# Tắt chế độ tính gradient khi chỉ cần in kết quả
with torch.no_grad():
    print("\n--- Kiểm tra lại các phương trình ---")
    print(f"A + B = {A.item() + B.item():.4f} (Kỳ vọng: 9)")
    print(f"C - D = {C.item() - D.item():.4f} (Kỳ vọng: 1)")
    print(f"A + C = {A.item() + C.item():.4f} (Kỳ vọng: 8)")
    print(f"B - D = {B.item() - D.item():.4f} (Kỳ vọng: 2)")
    
    # Kiểm tra tổng bình phương sai số
    loss = (A+B-9)**2 + (C-D-1)**2 + (A+C-8)**2 + (B-D-2)**2
    print(f"Sai số (loss) cuối cùng: {loss.item():.6f}")
```

### Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau:
### • Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
### • Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
### • Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
### • Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
# 1. Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
X = torch.arange(24)
print("Tensor X ban đầu:\n", X)
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)
print("\nReshape to 4 x 6:\n", X_4x6)
print("\nReshape to 2 x 12:\n", X_2x12)
print("\nReshape to 2 x 3 x 4:\n", X_2x3x4)
```

```python
# 2.Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
A = torch.randn(3, 4)
B = torch.randn(3, 4) + 0.1 #cộng 0.1 vào B để tránh mẫu số bằng 0
print("\nA + B")
print(A + B)
print("\nA * B")
print(A * B)
print("\nA / B")
print(A / B)
print("\nTrung bình theo cột của A\n", A.mean(dim=0))
print("\nĐộ lệch chuẩn theo cột của A\n", A.std(dim=0))
```

```python
# 3. Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
# autograd
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

# Thủ công
x_val = 2.0
manual_grad = 4 * x_val + 5  # dy/dx = 4x + 5
print(f"Đạo hàm tính bằng Autograd (x.grad): {x.grad.item()}")
print(f"Đạo hàm tính thủ công              : {manual_grad}")
```

```python
# 4. Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.
import numpy as np
polynomial = np.poly1d([2, -3, 5])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)

print("\n--- So sánh hệ số ---")
print("Hệ số gốc (đáp án)    : [2.0, -3.0, 5.0]")
print(f"Hệ số mô hình học được: [{w[0].item():.4f}, {w[1].item():.4f}, {w[2].item():.4f}]")
sai_so = torch.abs(w.squeeze() - torch.tensor([2.0, -3.0, 5.0]))
print(f"Sai số tuyệt đối      : {sai_so.detach().numpy()}")
```
