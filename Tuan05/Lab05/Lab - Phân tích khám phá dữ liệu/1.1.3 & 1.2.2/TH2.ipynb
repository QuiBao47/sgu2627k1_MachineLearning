```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 1. Nạp tập dữ liệu rượu đỏ vào DataFrame (thay tên file cho khớp với file bạn đã tải)
wine_data = pd.read_csv("winequality-red.csv")

# Kiểm tra sơ bộ dữ liệu
print(wine_data.head())
print(wine_data.shape)

# ==========================================
# PHẦN A: TRỰC QUAN HÓA VỚI THƯ VIỆN MATPLOTLIB
# ==========================================

# Sắp xếp dữ liệu theo chất lượng (quality) hoặc độ cồn (alcohol) giảm dần để lấy top
wine_sorted = wine_data.sort_values("alcohol", ascending=False).reset_index(
    drop=True
)

# Case 1 & 2: Biểu đồ cột cơ bản và nâng cao (Top 10 loại rượu có độ cồn cao nhất)
plt.figure(figsize=(12, 6))
x = wine_sorted.index[0:10].astype(str)
y = wine_sorted["alcohol"][0:10]

plt.bar(x, y, color="skyblue")
plt.title(
    "Top 10 Wines with the highest alcohol content", fontsize=15, fontweight="bold"
)
plt.xlabel("Index of Wine", fontsize=12)
plt.ylabel("Alcohol Percentage", fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.show()

# Case 3: Đa góc nhìn (So sánh 2 thuộc tính: Alcohol và Fixed Acidity)
fig, ax = plt.subplots(figsize=(16, 7))
x = wine_sorted.index[0:10].astype(str)
y1 = wine_sorted["alcohol"][0:10]
y2 = wine_sorted["fixed acidity"][0:10]

# Subplot 1: Alcohol
plt.subplot(1, 2, 1)
plt.bar(x, y1, color="salmon")
plt.xlabel("Index of Wine", fontsize=12)
plt.ylabel("Alcohol", fontsize=12)
plt.title("Top 10 Wines - Alcohol", fontsize=14)

# Subplot 2: Fixed Acidity
plt.subplot(1, 2, 2)
plt.bar(x, y2, color="lightgreen")
plt.xlabel("Index of Wine", fontsize=12)
plt.ylabel("Fixed Acidity", fontsize=12)
plt.title("Top 10 Wines - Fixed Acidity", fontsize=14)

plt.tight_layout()
plt.show()


# ==========================================
# PHẦN B: TRỰC QUAN HÓA VỚI THƯ VIỆN SEABORN
# ==========================================

# Case 1 & 2: Biểu đồ barplot nâng cao với Seaborn
plt.figure(figsize=(12, 6))
data_top10 = wine_sorted[0:10]
ax = sns.barplot(
    data=data_top10, x=data_top10.index, y="alcohol", palette="viridis"
)
ax.set_xlabel("Index of Wine", fontsize=15)
ax.set_ylabel("Alcohol Percentage", fontsize=15)
ax.set_title(
    "Top 10 Wines with the highest alcohol content (Seaborn)",
    fontsize=20,
    fontweight="bold",
)
plt.show()

# Case 3: Xem nhiều góc nhìn cùng lúc bằng Seaborn (Subplots)
fig, ax = plt.subplots(1, 2, figsize=(18, 6))
sns.set(font_scale=1.2)

# Góc nhìn 1: Alcohol theo Index
ax1 = sns.barplot(
    data=data_top10, x=data_top10.index, y="alcohol", ax=ax[0], palette="Blues_d"
)
ax1.set_xlabel("Index")
ax1.set_ylabel("Alcohol")
ax1.set_title("Top 10 - Alcohol Perspective")

# Góc nhìn 2: pH theo Index
ax2 = sns.barplot(
    data=data_top10, x=data_top10.index, y="pH", ax=ax[1], palette="Oranges_d"
)
ax2.set_xlabel("Index")
ax2.set_ylabel("pH Level")
ax2.set_title("Top 10 - pH Perspective")

plt.tight_layout()
plt.show()
```
